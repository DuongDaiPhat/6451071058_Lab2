## Bước 1 & 2: Chuẩn bị thư viện và tải dữ liệu `load_digits`

```python
%matplotlib inline
print("Dương Đại Phát - 6451071058")
import matplotlib
import matplotlib.pyplot as plt
from sklearn import datasets, tree, metrics
from sklearn.model_selection import train_test_split

# The digits dataset
digits = datasets.load_digits()
print("Số lượng mẫu:", len(digits.images), "| Kích thước mỗi ảnh:", digits.images[0].shape)
```

## Bước 3: Biểu diễn 4 mẫu dữ liệu huấn luyện tương ứng với 4 lớp 0, 1, 2, 3

```python
print("Dương Đại Phát - 6451071058")
_, axes = plt.subplots(1, 4)
images_and_labels = list(zip(digits.images, digits.target))
for ax, (image, label) in zip(axes, images_and_labels[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Training: %i' % label)
plt.show()
```

## Bước 4, 5, 6, 7: Làm phẳng ảnh 8x8 -> 1x64, chia 80/20 và huấn luyện cây ID3

```python
print("Dương Đại Phát - 6451071058")
# To apply a classifier on this data, we need to flatten the image, to
# turn the data in a (samples, feature) matrix:
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

# Create a classifier: a decision tree classifier (ID3 - entropy)
classifier = tree.DecisionTreeClassifier(criterion="entropy", random_state=33)

# Split data into train and test subsets (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2, shuffle=False)

# We learn the digits on the train part of the digits
classifier.fit(X_train, y_train)
```

## Bước 8 & 9: Dự đoán trên tập kiểm thử và biểu diễn một vài kết quả

```python
print("Dương Đại Phát - 6451071058")
# Now predict the value of the digit on the test part:
predicted = classifier.predict(X_test)

_, axes = plt.subplots(1, 4)
images_and_predictions = list(zip(X_test.reshape(-1, 8, 8), predicted))
for ax, (image, prediction) in zip(axes, images_and_predictions[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Prediction: %i' % prediction)
plt.show()
```

## Bước 10: Thông số đạt được của cây ID3 và Ma trận nhầm lẫn

```python
print("Dương Đại Phát - 6451071058")
print("Classification report for classifier %s:\n%s\n"
      % (classifier, metrics.classification_report(y_test, predicted)))

disp = metrics.ConfusionMatrixDisplay.from_estimator(classifier, X_test, y_test)
disp.figure_.suptitle("Confusion Matrix")
print("Confusion matrix:\n%s" % disp.confusion_matrix)
plt.show()
```

## Bước 11: Nhận dạng chữ số viết tay từ ảnh chụp thực tế (`test_image_Bai7.jpg`)

```python
print("Dương Đại Phát - 6451071058")
from PIL import Image, ImageOps
import numpy as np

# Open image
img = Image.open('test_image_Bai7.jpg').convert("L").resize((8, 8))
# Invert image
img = ImageOps.invert(img)
# Convert to numpy array
im2arr = np.array(img)
# Show
plt.imshow(im2arr, cmap=plt.cm.gray_r, interpolation='nearest')
plt.title("Ảnh sau khi chuyển mức xám 8x8 và đảo màu")
plt.show()
```

```python
print("Dương Đại Phát - 6451071058")
# Change from 2d array (8x8) to 1d array (1x64)
img1d = im2arr.reshape([1, 64])
# Normalize black & white image array
img1d[img1d > 109] = 155
img1d[img1d < 110] = 0
print("Mảng 1x64 sau chuẩn hóa:\n", img1d)

# After normalize
plt.imshow(img1d.reshape(8, 8), cmap=plt.cm.gray_r, interpolation='nearest')
plt.title("Hình ảnh chữ số sau khi làm rõ (ngưỡng 110)")
plt.show()

# Predict label of number
y_pred = classifier.predict(img1d)
print("Kết quả nhận diện chữ số:", y_pred)
```
